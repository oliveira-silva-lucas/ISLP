# LDA Linear Drisciminant Analysis #

δk(x) = xᵀ Σˉ¹ µk − 1/2 µkᵀ Σˉ¹ µk + log(πk)

```python
lda = LDA(store_covariance=True)
```

```python
#Remover Intercepto pois LDA adiciona
X_train, X_test = [M.drop(columns=['intercept']) 
                   for M in [X_train, X_test]]
#Como é LDA podemos passar a informação de 'Direction' como y da Predição, k -> Direction-> ['Down', 'Up']
lda.fit(X_train, L_train)
```

µk - vetor de média k para cada classe

```python
lda.means_
```

```python
lda.classes_
```

Probabilidade a Priori das classes

```python
print(f'πDown:{lda.priors_[0]}\nπUp:{lda.priors_[1]}')
lda.priors_
```

```python
lda.scalings_
```

```python
lda_pred = lda.predict(X_test)
```

```python
#Resultado já vem em Valor de Classe k
confusion_table(lda_pred, L_test)
```

```python
print(f'Probabilidade nos dados de teste:\nAcerto: {np.mean(lda_pred==L_test)}\nErro: {np.mean(lda_pred!=L_test)}')
```

Devolver resultado da predição em probabilidade

```python
lda_prob = lda.predict_proba(X_test)
lda_prob[:12]
```

```python
#Transformando e comparando se resultados foram iguais
np.all(
    np.where(lda_prob[:,1]>=0.5, 'Up', 'Down') == lda_pred
)
```

```python
#Caso seja k>2
np.all(
    #np.argmax devolve o índice do item com maior valor
    [lda.classes_[i] for i in np.argmax(lda_prob, 1)] == lda_pred
)
```

Se quiser utilizar um limiar/threshold diferente de 0.5

```python
print('Limiar padrão 0.5:\n',pd.Series(lda_pred).value_counts())
print()
lda_limiar_51 = np.array(['Down']*len(lda_prob))
lda_limiar_51[lda_prob[:,0]>0.51] = 'Up'
print('Limiar customizado 0.51:\n', pd.Series(lda_limiar_51).value_counts())
```

# QDA Quadratic Discriminant Analysis #

δk(x) = -1/2 (x-µk)ᵀ Σkˉ¹ (x-µk)ᵀ − 1/2 log|Σk| + log(πk)

```python
qda = QDA(store_covariance=True)
qda.fit(X_train, L_train)
```

```python
#vetores de média e probabilidade a priori de cada classe
qda.means_, qda.priors_
```

Matrizes de covariância para cada K, diferentemente do LDA que é uma matriz de covariância para todos K, no QDA cada K tem a sua própria matriz de covariância calculada a partir das observações pertecentes a cada classe k.

```python
qda.covariance_
```

```python
qda_pred = qda.predict(X_test)
confusion_table(qda_pred, L_test)
```

```python
print(f'Probabilidade nos dados de teste:\nAcerto: {np.mean(qda_pred==L_test)}\nErro: {np.mean(qda_pred!=L_test)}')
```

```python
#Para resultados ser retornado em probabilidade segue o mesmo principio de LDA
qda.predict_proba(X_test)
```

# Caravan Dataset #

```python
Caravan = load_data('Caravan')
Purchase = Caravan.Purchase
Purchase.value_counts()
```

```python
Purchase.value_counts()[1]/len(Purchase)
```

```python
feature_df = Caravan.drop(columns=['Purchase'])
```

```python
scaler = StandardScaler(with_mean=True,
                     with_std=True,
                     copy=True)
scaler.fit(feature_df)
X_std = scaler.fit_transform(feature_df)
feature_std = pd.DataFrame(X_std, columns=feature_df.columns)
feature_std.std()
```

```python
(X_train, 
 X_test, 
 y_train, 
 y_test) = train_test_split(feature_std,
                                                      Purchase,
                                                      test_size=1000,
                                                      random_state=0)
```

## KNN ##

```python
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1_pred = knn1.fit(X_train, y_train).predict(X_test)
print(f'Probabilidade nos dados de Teste:\nAcerto: {np.mean(knn1_pred == y_test)*100}%\nErro: {np.mean(knn1_pred != y_test)*100}%')
print(np.mean(knn1_pred != y_test), np.mean(y_test!='No'))
confusion_table(knn1_pred, y_test)
```

```python
print(f'Probabilidade de acertar corretamente quem compra(Purchase): {(9/(53+9))*100}%')
```

Procurar melhor quantidade de K para KNN com iteração

```python
for K in range (1, 5):
    knn = KNeighborsClassifier(n_neighbors=K)
    knn_pred = knn.fit(X_train, y_train).predict(X_test)
    C = confusion_table(knn_pred, y_test)
    templ = ('K={0:d}: # Predito como Positivo: {1:>2} # Verdadeiro Positivo {2:d}, Precisão {3:.1%}')
    pred = C.loc['Yes'].sum()
    did_rent = C.loc['Yes','Yes']
    print(templ.format(
        K,
        pred,
        did_rent,
        did_rent/pred
    ))
```

## Logistic Regression ##

```python
logit = LogisticRegression(C=1e10, solver='liblinear')
logit.fit(X_train, y_train)
logit_pred = logit.predict_proba(X_test)
logit_labels = np.where(logit_pred[:,1] > .5, 'Yes', 'No')
print(f'Probabilidade nos dados de Teste:\nAcerto: {np.mean(logit_labels == y_test)*100}%\nErro: {np.mean(logit_labels != y_test)*100}%')
C = confusion_table(logit_labels, y_test)
templ = ('Limiar={0}: # Predito como Positivo: {1:>2} # Verdadeiro Positivo {2:d}, Precisão {3:.1%}')
pred = C.loc['Yes'].sum()
did_rent = C.loc['Yes','Yes']
print(templ.format(
    0.5,
    pred,
    did_rent,
    did_rent/pred
))
C
```

```python
logit_labels = np.where(logit_pred[:,1] > .25, 'Yes', 'No')
C = confusion_table(logit_labels, y_test)
pred = C.loc['Yes'].sum()
did_rent = C.loc['Yes','Yes']
print(templ.format(
    0.25,
    pred,
    did_rent,
    did_rent/pred
))
C
```

# Bikeshare Dataset #

```python
Bike = load_data('Bikeshare')
```

```python
Bike.shape, Bike.columns
```

```python
X = MS(['mnth',
        'hr',
        'workingday',
        'temp',
        'weathersit']).fit_transform(Bike)
Y = Bike['bikers']
```

## Linear Regression ##

```python
M_lm = sm.OLS(Y, X).fit()
#Mês de janeiro e hora 0 foram tratados como valor base e estão no intercepto
summarize(M_lm)
```

```python
hr_econde = contrast('hr', 'sum')
mnth_encode = contrast('mnth', 'sum')
```

```python
#Agora dezembro e hora 23 foram removidos, a última opção
X2 = MS([mnth_encode,
         hr_econde,
         'workingday',
         'temp',
         'weathersit']).fit_transform(Bike)
M2_lm = sm.OLS(Y, X2).fit()
S2 = summarize(M2_lm)
S2
```

A diferença é apenas de interpretação a qualidade do modelo não muda

```python
#Com a mudança de qual opção das classes está no intercepto o valor delas se diferenciam
M_lm.params['intercept'], M2_lm.params['intercept']
```

```python
np.sum((M_lm.fittedvalues - M2_lm.fittedvalues)**2)
```

```python
np.allclose(M_lm.fittedvalues, M2_lm.fittedvalues)
```

```python
np.sum(M_lm.rsquared), np.sum(M2_lm.rsquared)
```

```python
coef_mnth = S2[S2.index.str.contains('mnth')]['coef']
coef_mnth
```

```python
months = Bike['mnth'].dtype.categories
coef_mnth = pd.concat([
    coef_mnth,
    #podemos calcular o coeficiente da opção base fazendo o cálculo a seguir, só funciona por causa do contrast anteriormente chamado
    pd.Series([-coef_mnth.sum()],
              index=['mnth[Dec]'])
])
coef_mnth
```

```python
fig_month, ax_month = subplots(figsize=(8,8))
x_month = np.arange(coef_mnth.shape[0])
ax_month.plot(x_month, coef_mnth, marker='o', ms=10)
ax_month.set_xticks(x_month)
ax_month.set_xticklabels([l[5] for l in coef_mnth.index], fontsize=20)
ax_month.set_xlabel('Month', fontsize=20)
ax_month.set_ylabel('Coefficient', fontsize=20);
```

```python
coef_hr = S2[S2.index.str.contains('hr')]['coef']
coef_hr = coef_hr.reindex(['hr[{0}]'.format(h) for h in range(23)])
coef_hr = pd.concat([
    coef_hr,
    pd.Series([-coef_hr.sum()], index=['hr[23]'])
])
```

```python
fig_hr, ax_hr = subplots(figsize=(8,8))
x_hr = np.arange(coef_hr.shape[0])
ax_hr.plot(x_hr, coef_hr, marker='o', ms=10)
ax_hr.set_xticks(x_hr[::2])
ax_hr.set_xticklabels(range(24)[::2], fontsize=20)
ax_hr.set_xlabel('Hour', fontsize=20)
ax_hr.set_ylabel('Coefficient', fontsize=20);
```

## Poisson Regression ##


```python
M_pois = sm.GLM(Y, X, family=sm.families.Poisson()).fit()
```

```python
S_pois = summarize(M_pois)
coef_month = S_pois[S_pois.index.str.contains('mnth')]['coef']
coef_month = pd.concat([
    coef_month,
    pd.Series([-coef_month.sum()],
              index=['mnth[Dec]'])
])
coef_hr = S_pois[S_pois.index.str.contains('hr')]['coef']
coef_hr = pd.concat([coef_hr,
                     pd.Series([-coef_hr.sum()], index=['hr[23]'])])
```

```python
fig_pois, (ax_month, ax_hr) = subplots(1, 2, figsize=(16,8))
ax_month.plot(x_month, coef_month, marker='o', ms=10)
ax_month.set_xticks(x_month)
ax_month.set_xticklabels([l[5] for l in coef_month.index], fontsize=20)
ax_month.set_xlabel('Month', fontsize=20)
ax_month.set_ylabel('Coefficient', fontsize=20)
ax_hr.plot(x_hr, coef_hr, marker='o', ms=10)
ax_hr.set_xticklabels(range(24)[::2], fontsize=20)
ax_hr.set_xlabel('Hour', fontsize=20)
ax_hr.set_ylabel('Coefficient', fontsize=20);
```

```python
fig, ax = subplots(figsize=(8, 8))
ax.scatter(M2_lm.fittedvalues, M_pois.fittedvalues, s=20)
ax.set_xlabel('Linear Regression Fit', fontsize=20)
ax.set_ylabel('Poisson Regression Fit', fontsize=20)
ax.axline([0,0], c='black', linewidth=3, linestyle='--', slope=1);
```
