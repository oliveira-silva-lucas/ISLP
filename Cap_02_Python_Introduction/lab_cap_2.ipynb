```python
A = np.array(np.arange(16)).reshape((4,4))
A
```

```python
#Selecionar um item específico
A[1, 2]
```

```python
#Selecionar linhas
lista_linhas = [1, 3]
A[lista_linhas]
```

```python
#Passando a lista de linhas de forma direta
A[[1,3]]
```

```python
#Fazer o mesmo com colunas
lista_colunas = [0, 2]
A[:, lista_colunas]
```

```python
A[:,[0, 2]]
```

Selecionar linhas e colunas de uma matriz

```python
#por dedução lógica das for anteriores, lista de linha de pois lista de coluna temos um resultado
#não esperado
print(A[lista_linhas, lista_colunas])
#ou
A[[1, 3], [0, 2]]
```

```python
#o resultado veio como se tivessemos selecionado dois pontos específicos da matriz
np.array([A[1, 0], A[3, 2]])
```

```python
#se tentarmos selecionar da mesma forma, duas linhas, com a informação de 3 colunas temos erro
A[[1,3], [0, 2, 3]]
```

A solução é compreender a forma que geramos a submatriz, fazemos uma, depois a outra

```python
print(A[lista_linhas][:, lista_colunas])
#ou
print("\n", A[:, lista_colunas][lista_linhas])
#ou
print("\n", A[[1, 3]] [:,[0, 2]])
#ou
print("\n", A[:,[0, 2]] [[1, 3]])
```

```python
A[[1, 3]][:, [0, 2, 3]]
```

Forma mais legível de buscar uma submatriz com np.ix

```python
idx = np.ix_([1,3], [0, 2, 3])
A[idx]
```

Forma mais prática de buscar submatriz com slices

```python
print(A)
A[1:4:2, 0:3:2]
```

```python
keep_rows = np.zeros(A.shape[0], bool)
keep_rows
```

```python
keep_rows[[1,3]] = True
keep_rows
```

```python
#Boleanos tem valores que na comparação são (0,1)
np.all( keep_rows == np.array([0,1,0,1]))
```

```python
print(A)
A[np.array([0,1,0,1])]
```

```python
print(A)
#Quando passamos uma lista bolleana o numpy entende como filtro de exibição de linhas
A[keep_rows]
```

```python
keep_cols = np.zeros(A.shape[1], bool)
keep_cols[[0,2,3]] = True
idx_bool = np.ix_(keep_rows, keep_cols)
A[idx_bool]
```

```python
idx_mixed = np.ix_([1,3], keep_cols)
A[idx_mixed]
```

```python
total = 0
for value in [3, 2, 19]:
    total += value
print("total is: {0}".format(total))
```

```python
total = 0
for value in [3, 2, 19]:
    for weight in [3, 2, 1]:
        total += value * weight
print("total is: {0}".format(total))
```

```python
total = 0
for value, weight in zip([2, 3, 19],
                         [0.2, 0.3, 0.5]):
    total += weight * value

print("weighted avarege is: {0}".format(total))
```
