```python
import pandas as pd

college = pd.read_csv("College.csv")
college
```

```python
college2 = pd.read_csv("College.csv", index_col=0)
college2
```

```python
college3 = college.rename({"Unnamed: 0": 'College'}, axis=1)
college3 = college3.set_index("College")
college3
```

```python
college = college3
```

```python
college.describe()
```

```python
c = ["Top10perc","Apps","Enroll"]
pd.plotting.scatter_matrix(college[c]);
```

```python
college.boxplot("Outstate", by="Private");
```

```python
college["Elite"] = pd.cut(college['Top10perc']/100,
       [0, 0.5, 1],
       labels=["No", "Yes"])
college["Elite"].value_counts()
```

```python
college.boxplot('Outstate', by='Elite')
```

```python
college.hist(figsize=(12,6), bins=60);
```
