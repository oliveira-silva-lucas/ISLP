(c) Now estimate the standard error of ˆµ using the bootstrap. How
does this compare to your answer from (b)?

```python
def boot_SE(func,
            D, 
            n=None,
            B=1000,
            seed=0):
    rng = np.random.default_rng(seed)
    first_, second_ = 0, 0
    n = n or D.shape[0]
    for _ in range(B):
        idx = rng.choice(D.index,
                         n,
                         replace=True)
        value = func(D, idx)
        first_ += value
        second_ += value ** 2
    return np.sqrt(second_/B - (first_/B)**2), first_/B
```

```python
def boot_media (D, idx):
    return np.mean(D.loc[idx])
boot_erro_padrao, boot_medv_media = boot_SE(func=boot_media, D=boston.medv, seed=42)
boot_erro_padrao, boot_medv_media
```
