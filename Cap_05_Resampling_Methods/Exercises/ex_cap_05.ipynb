2.(g) Create a plot that displays, foreach integer value of n from 1
to 100,000, the probability that the jth observation is in the
bootstrap sample.Commenton what you observe.

```python
def calc_probabilidade_bootstrap(n):
    return 1-(1 -1/n)**n

for i in range(1,6):
    print(f'número {i}, probabilidade de {calc_probabilidade_bootstrap(i)}')
```

```python
x =  np.arange(1, 100_001)
y = calc_probabilidade_bootstrap(x)

x[:10], y[:10]
```

```python
fig, ax = plt.subplots(figsize=(6,6))
ax.plot(x,y)
ax.set_xscale('log')
ax.set_ylabel('Probabilidade')
ax.set_xlabel('Número')
```

a medida que o número aumenta sua probabilidade vai se concentrando em torno de 0.63

(h) We will now investigate numerically the probability that a bootstrap sample of size n=100 contains the jth observation.Here j=4. We first create an array store with values that will subsequently be overwritten using the function np.empty().We then repeatedly create bootstrap samples, and each time we record
whether or not the fifth observation is contained in the bootstrap
sample.

```python
rng = np.random.default_rng(42)
store = np.empty(100_000)
for i in range(100_000):
    store[i] = np.sum(rng.choice(100, size=100, replace=True) == 4)>0

np.mean(store)
```

Como esperado dado o entendimento do exercicio anterior, a probabilidade do número 4 ocorrer está em torno do 0.63...

Entendimento do exercício:

Bootstrap é uma técnica onde a partir de um conjunto/dataset se cria n subconjuntos de mesmo tamanho, porém repetindo alguns elementos e excluindo outros, dessa forma avaliamos a variabilidade dos subconjuntos e com isso podemos calcular a variabilidade, erro padrão, estabilidade de um preditor e/ou modelo esperado em um novo conjunto desconhecido.

Neste exercicio foi simulado que o conjunto/dataset padrão é um array de 0 a 99, a cada iteração do 'for' sorteamos um conjunto aleatório de 100 itens que podem conter os valores de 0 a 99, ao fazer essa iteração por uma quantidade alta chegamos a uma estabilização da probabilidade de um dado do conjunto padrão estar contido nos subconjuntos.
